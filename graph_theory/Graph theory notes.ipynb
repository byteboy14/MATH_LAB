## Isomorphic graph

### Equality of Graphs

Two graphs $G$ and $H$ are equal if and only if

$$
G = H \iff V(G) = V(H), \quad E(G) = E(H)
$$

and their edge functions agree:

$$
\psi_G(e) = \psi_H(e)
\quad \text{for every } e \in E(G).
$$

### Graph Isomorphism

Two graphs $G$ and $H$ are isomorphic if and only if there exist bijections

$$
\theta : V(G) \to V(H)
$$

and

$$
\phi : E(G) \to E(H)
$$

such that, for every edge $e \in E(G)$, if

$$
\psi_G(e) = uv,
$$

then

$$
\psi_H(\phi(e)) = \theta(u)\theta(v).
$$

### Complete Bipartite Graph

A **simple bipartite** is one whose vertex set can be partitioned into two subsets X and Y,
so that each edge has one end in X and one end in Y.
A simple bipartite graph with bipartition $(X,Y)$ in which each vertex of $X$ is joined to each vertex of $Y$ is called a **complete bipartite graph**.

If $|X|=m$ and $|Y|=n$, it is denoted by $K_{m,n}$.

**Proposition** : Let $G$ be a simple graph with $n$ vertices and no edges. Prove that $G$ is isomorphic to $K_{n,0}$.
