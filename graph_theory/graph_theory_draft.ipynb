## Bipartite K4,4

```python
G = graphs.CompleteBipartiteGraph(4, 4)
G.show()
```

```python
A, B = G.bipartite_sets()

print(A)
print(B)
```

```python
G1 = G.subgraph(A)
G2 = G.subgraph(B)

G1.show()
G2.show()
```
