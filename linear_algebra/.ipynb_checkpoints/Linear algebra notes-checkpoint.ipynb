# LINEAR ALGEBRA NOTES

Let $ u , v  $ vectors 

Linear combination of $ u,v $ is $ cu +dv $ with $c,d$ scalars

## Linear combinations

Note , all vector by default are columns vectors.

In 3D space :

+ $ cu $ : gave a line through zero vector (0,0,0)

+ $cu + du $ : gave a plane through zero vector (0,0,0)

+ $ cu + du + ew $ : gave whole space .

This depend of the dimension of $ u, v , w $

### Ex : ### 
- The linear combinations of v = (1,1,0) and w = (0,1,1) fill a plane in R 3 .
Describe that plane. Find a vector that is not a combination of v and w-not on the plane.

- For v = (1, 0) and w = (0, 1), describe all points $ cv $ with (1) whole numbers $ c $
(2) nonnegative numbers $ c \geq 0 $. Then add all vectors $ dw $ and describe all $ cv + dw $.

-  Find two equations for $ c $ and $ d $ so that the linear combination $ cv + dw = b $ :

$ v = \left[\begin{array}{rrr}
2 \\
-1
\end{array} \right]
$ , 
$ w = \left[\begin{array}{rrr}
-1 \\
2
\end{array} \right]
$ , 
$ b = \left[\begin{array}{rrr}
1 \\
0
\end{array} \right]
$
