```python
f(x) = exp(x) * cos(x); f(x) ; z = 0 ; xi = function('xi') ;
```

```python
P2(x) = f(z) + diff(f,x,1).subs(x=z)*(x-z) + diff(f,x,2).subs(x=z)*(x-z)^2/2 ; P2
```

```python
P(x) = P2(x) ; P
```

```python
y = 3; R(x) = diff(f,x,y).subs(x=xi(x))*(x-z)^y/factorial(y) ; R
```

```python
M(x) = 2*exp(x) * (cos(x) - sin(x)) ; M # find M to bound xi value
```

```python
y = 3 ; Rr(x) = M*(x-z)^y/factorial(y); Rr
```

```python
Rs(x) = f(x) - P(x) ; Rs
```

```python
m = 0.2
```

```python
Er = Rr(m).n() ; Er
```

```python
E = Rs(m) ; E.n()
```

```python
f(m).n() == P(m)
```

```python
f(m).n() == P(m) + Er
```

```python
plot(Rr, (x, 0, 1), color='red') + plot(Rs, (x, 0, 1))
```

```python
max = 5
@interact
def graph(max=slider(1, 5*max, 1, default=max)):
    show(
        plot(Rr, (x, 0, max), color='red')
        + plot(Rs, (x, 0, max))
    )
```
