```python
f(x) = sqrt(x+1); f(x) ; z = 0 ; xi = function('xi') ;
```

```python
P2(x) = f(z) + diff(f,x,1).subs(x=z)*(x-z) + diff(f,x,2).subs(x=z)*(x-z)^2/2 ; P2
```

```python
P(x) = P2(x) + diff(f,x,3).subs(x=z)*(x-z)^3/factorial(3) ; P
```

```python
taylor(f,x,z,3)
```

```python
R(x) = diff(f,x,4).subs(x=xi(x))*(x-z)^4/factorial(4) ; R
```

```python
E(x) = f(x) - P(x); E
```
