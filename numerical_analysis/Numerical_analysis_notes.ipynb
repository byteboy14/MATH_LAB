# Taylor Polynomials and Series

## Taylor's Theorem

$$
f(x) = P_n(x) + R_n(x)
$$

Where



$$
P_n(x) = \sum_{k=0}^{n} \frac{f^{(k)}(a)}{k!}(x-a)^k
$$

and 


$$
R_n(x) = \frac {f^{(n+1)}(\xi{(x)})}{(n+1)!} (x-x_0)^{(n+1)}
$$

with $ \xi(x) \leq |x-x_0| $
