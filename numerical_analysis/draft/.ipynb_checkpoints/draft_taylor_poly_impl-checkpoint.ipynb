# Taylor's Theorem

$$
f(x) = P_n(x) + R_n(x)
$$



# Taylor's Theorem $P(x)$



$$
P_n(x) = \sum_{k=0}^{n} \frac{f^{(k)}(x_0)}{k!}(x-x_0)^k
$$

```python
def k_taylor(f,x0,k) :
    return (diff(f,x,k)(x0) * (x-x0)**k)/factorial(k);
    
```

```python
def mTaylor(f,x0,n) :
    var('k')
    p = 0;
    for k in range(0,n+1) :
        p += k_taylor(f,x0,k);

    P(x) = p ;
    return P;
    
```

```python
x0 = 1
n = 3
f(x) = x**30 ; f
```

```python
mTaylor(f,x0,n)
```

```python
f.taylor(x, x0, n)
```

```python
bool((mTaylor(f,x0,n) - f.taylor(x, x0, n)).simplify_full() == 0)
```

## Taylor's Theorem $R(x)$

$$
R_n(x) = \frac {f^{(n+1)}(\xi{(x)})}{(n+1)!} (x-x_0)^{(n+1)}
$$
