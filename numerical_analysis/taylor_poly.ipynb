```python
f(x) = x^3; f(x)
```

```python
z = 0; P(x) = f(z) + diff(f,x,1).subs(x=z)*(x-z) + diff(f,x,2).subs(x=z)*(x-z)^2/2 ; P
```

```python
taylor(f,x,z,2)
```

```python
xi = function('xi') ; xi(x)
```

```python
R(x) = diff(f,x,3).subs(x=xi(x))*(x-z)^3/6 ; R
```

```python
R(0.5)
```

```python
E(x) = f(x) - P(x); E
```

```python
E(0.5)
```

```python
E(0.5) == R(0.5)
```
