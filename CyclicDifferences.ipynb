<h2>Experimentation géometrique du système de différence cyclic</h2>

```python
# Paramètre
var('t')

# Paramétrisation
x1 = sin(t)
x2 = cos(t)
x3 = cos(2*t)

# Vecteur dépendant de t
v = vector([
    x1 - x3,
    x2 - x1,
    x3 - x2
])

# Domaine du paramètre
T = srange(0, 2*pi, 0.1)

# Création des frames
frames = []

for tt in T:
    vt = v.subs(t=tt).n()
    frames.append(
        arrow3d(
            (0, 0, 0),
            vt,
            color='blue',
            width=3
        )
    )

# Animation
anim = animate(frames)
anim
```

```python
# Création des frames
frames = []

for tt in T:
    vt = v.subs(t=tt).n()
    
    # Normalisation (sécurité incluse)
    if vt.norm() != 0:
        vt_unit = vt / vt.norm()
    else:
        vt_unit = vt
    
    frames.append(
        arrow3d(
            (0, 0, 0),
            vt_unit,
            color='blue',
            width=3
        )
    )

# Animation
anim = animate(frames)
anim.show()
```

```python
# Création des frames
frames = []

for tt in T:
    vt = v.subs(t=tt).n()
        
    frames.append(
        trajectory +
        arrow3d(
            (0, 0, 0),
            vt,
            color='blue',
            width=3
        )
    )

# Animation
anim = animate(frames)
anim.show()
```
